```python
import asyncio
import threading
import time

# ASYNC APPROACH - Efficient for I/O
async def async_benchmark():
    start = time.time()
    
    async def read_device(port_num):
        # Simulate serial read
        await asyncio.sleep(0.1)  # Network I/O
        return f"Data from port {port_num}"
    
    # Handle 50 devices
    results = await asyncio.gather(*[
        read_device(i) for i in range(50)
    ])
    
    end = time.time()
    print(f"Async: {end - start:.2f}s for 50 devices")

# THREADING APPROACH - Heavy for I/O
def threading_benchmark():
    start = time.time()
    results = []
    lock = threading.Lock()
    
    def read_device(port_num):
        time.sleep(0.1)  # Blocking I/O
        with lock:
            results.append(f"Data from port {port_num}")
    
    threads = []
    for i in range(50):
        t = threading.Thread(target=read_device, args=(i,))
        threads.append(t)
        t.start()
    
    for t in threads:
        t.join()
    
    end = time.time()
    print(f"Threading: {end - start:.2f}s for 50 devices")

# Run the benchmarks
threading_benchmark()
await async_benchmark()  # Use 'await' in Jupyter notebook
```
